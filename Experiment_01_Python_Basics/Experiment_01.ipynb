## 1. Python Tokens, Variables and Data Types

### 1(a) Python Tokens and Keywords

**Python tokens** are the smallest meaningful units of a Python program.  
They include **keywords, identifiers, literals, operators and delimiters**.

The following program displays Python's reserved keywords and demonstrates different types of tokens.

```python
import keyword
student_name = "Prerita"
experiment_no = 1
is_completed = True

print("PYTHON TOKEN DEMONSTRATION") 
print("-" * 35)
print("Identifier:", student_name)
print("Integer Literal:", experiment_no)
print("Boolean Literal:", is_completed)

print("\nTotal number of Python Keywords:", len(keyword.kwlist))
print("\nPython Keywords:")
print(keyword.kwlist)
```

### 1(b) Variable Declaration and Dynamic Typing

Python does not require the data type of a variable to be declared explicitly.  
The type is determined automatically at runtime and can change during program execution.

```python
# demonstration of Variable declaration and Dynamic Typing

value = 25
print("Value:", value, "| Type:", type(value))

value = 25.75
print("Value:", value, "| Type:", type(value))

value = "Python Programming"
print("Value:", value, "| Type:", type(value))

value = True
print("Value:", value, "| Type:", type(value))
```

### 1(c) Basic Data Types

Python provides several built-in data types.  
The following program demonstrates the four basic types: **integer, float, string and boolean**.

```python
# Demonstration of Basic Data Types

student_age = 18
attendance_percentage = 92.5
course_name = "Python Programming"
lab_completed = True

print("BASIC DATA TYPES")
print("-" * 35)

print("Student Age :", student_age,
      "→", type(student_age).__name__)

print("Attendance  :", attendance_percentage,
      "→", type(attendance_percentage).__name__)

print("Course Name :", course_name,
      "→", type(course_name).__name__)

print("Lab Status  :", lab_completed,
      "→", type(lab_completed).__name__)
```

## 2. Python Operators

Operators are symbols used to perform operations on variables and values.  
The following programs demonstrate **arithmetic, relational, logical and assignment operators**.

```python
# Demonstration of Different Types of Operators

a = 20
b = 6

print("1. ARITHMETIC OPERATORS")
print("-" * 35)
print("Addition       :", a + b)
print("Subtraction    :", a - b)
print("Multiplication :", a * b)
print("Division       :", a / b)
print("Floor Division :", a // b)
print("Modulus        :", a % b)
print("Exponentiation :", a ** 2)


print("\n2. RELATIONAL OPERATORS")
print("-" * 35)
print("a > b  :", a > b)
print("a < b  :", a < b)
print("a == b :", a == b)
print("a != b :", a != b)
print("a >= b :", a >= b)
print("a <= b :", a <= b)


print("\n3. LOGICAL OPERATORS")
print("-" * 35)

x = True
y = False

print("x and y :", x and y)
print("x or y  :", x or y)
print("not x   :", not x)


print("\n4. ASSIGNMENT OPERATORS")
print("-" * 35)

score = 10
print("Initial Score :", score)

score += 5
print("After += 5    :", score)

score -= 3
print("After -= 3    :", score)

score *= 2
print("After *= 2    :", score)

score /= 4
print("After /= 4    :", score)
```

## 3. Operator Precedence

Operator precedence determines the order in which operations are performed in an expression.  
Parentheses can be used to change the default order of evaluation.

```python
# Demonstration of Operator Precedence

result_1 = 10 + 5 * 2
result_2 = (10 + 5) * 2
result_3 = 2 ** 3 * 4
result_4 = 20 / 5 + 3 * 2

print("OPERATOR PRECEDENCE")
print("-" * 35)

print("10 + 5 * 2       =", result_1)
print("(10 + 5) * 2     =", result_2)
print("2 ** 3 * 4       =", result_3)
print("20 / 5 + 3 * 2   =", result_4)
```

## 4. Type Casting

Type casting is the process of converting a value from one data type to another.  
Python provides functions such as `int()`, `float()`, `str()` and `bool()` for type conversion.

```python
# Demonstration of Type Casting

num_text = "50"
decimal_num = 25.75
whole_num = 15
status = 1

print("TYPE CASTING")
print("-" * 35)

print("Original value :", num_text,
      "| Type:", type(num_text).__name__)

converted_int = int(num_text)
print("String to Integer :", converted_int,
      "| Type:", type(converted_int).__name__)

converted_float = float(whole_num)
print("Integer to Float  :", converted_float,
      "| Type:", type(converted_float).__name__)

converted_string = str(decimal_num)
print("Float to String   :", converted_string,
      "| Type:", type(converted_string).__name__)

converted_bool = bool(status)
print("Integer to Boolean:", converted_bool,
      "| Type:", type(converted_bool).__name__)
```

## 5. User Input and Basic Computation

The `input()` function is used to accept data from the user.  
The following program accepts student details and marks, then calculates the total and percentage.

```python
# Student Performance Calculator using input()

print("STUDENT PERFORMANCE CALCULATOR")
print("-" * 40)

name = input("Enter student name: ")

python_marks = float(input("Enter Python marks (out of 100): "))
maths_marks = float(input("Enter Mathematics marks (out of 100): "))
physics_marks = float(input("Enter Physics marks (out of 100): "))

total = python_marks + maths_marks + physics_marks
percentage = total / 3

print("\n--- RESULT ---")
print("Student    :", name)
print("Total      :", total, "/ 300")
print("Percentage :", round(percentage, 2), "%")
```
