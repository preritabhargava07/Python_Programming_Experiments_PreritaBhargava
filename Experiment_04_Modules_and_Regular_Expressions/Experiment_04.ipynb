## 3. Creating a Python Package with Multiple Modules

A Python package is a collection of related modules organized within a directory.

### Program: Student Academic Toolkit

A custom package named `student_toolkit` is created using two modules:

1. **academic.py** — Calculates total marks and average.
2. **attendance.py** — Calculates attendance percentage and checks eligibility.

## 4. Regular Expressions

Regular Expressions (Regex) are patterns used to search, match and validate text.

Python provides the built-in `re` module for working with regular expressions.

### 4(a) Email Address Validation

**Aim:** To validate email addresses using regular expressions.

**Concept:** The `re.fullmatch()` function checks whether the entire string matches a specified pattern.

```python
# Experiment 4 - Question 4(a)
# Email Address Validation using Regular Expressions

import re

def validate_email(email):
    pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9-]+(?:\.[A-Za-z0-9-]+)+"
    return bool(re.fullmatch(pattern, email))


emails = [
    "neo.student@gmail.com",
    "python_lab@university.edu",
    "invalid-email.com",
    "student@",
    "hello@college"
]

print("EMAIL VALIDATION SYSTEM")
print("-" * 45)

for email in emails:
    status = "Valid" if validate_email(email) else "Invalid"
    print(f"{email:<30} : {status}")
```

### 4(b) Mobile Number Extraction

**Aim:** To extract valid Indian mobile numbers from a given text using regular expressions.

**Concept:** The `re.findall()` function searches a string and returns all matching patterns.

The program identifies 10-digit Indian mobile numbers beginning with digits 6–9.

```python
# Experiment 4 - Question 4(b)
# Extracting Indian Mobile Numbers using Regex

import re

text = """
Student Contact Directory:
Aarav: 9876543210
Riya: 8765432109
Kabir: 7654321098
Invalid Contact: 1234567890
Emergency: 112
"""

pattern = r"(?<!\d)[6-9]\d{9}(?!\d)"

mobile_numbers = re.findall(pattern, text)

print("MOBILE NUMBER EXTRACTION")
print("-" * 45)

print("Extracted Mobile Numbers:")

for index, number in enumerate(mobile_numbers, start=1):
    print(f"{index}. {number}")
```

### 4(c) Demonstration of Regular Expression Metacharacters

**Aim:** To understand the use of metacharacters in regular expressions.

Metacharacters are special characters that define how a pattern should be matched.

| Metacharacter | Meaning |
|---|---|
| `.` | Matches any character except newline by default |
| `^` | Matches the beginning of a string |
| `$` | Matches the end of a string |
| `*` | Matches zero or more repetitions |
| `+` | Matches one or more repetitions |
| `?` | Matches zero or one repetition |
| `\d` | Matches a digit |
| `\w` | Matches a word character |
| `\s` | Matches whitespace |

```python
# Experiment 4 - Question 4(c)
# Demonstration of Regex Metacharacters

import re

print("REGEX METACHARACTERS DEMONSTRATION")
print("-" * 45)

examples = [
    (r"c.t", "cat"),
    (r"^Python", "Python is powerful"),
    (r"end$", "This is the end"),
    (r"ab*c", "abbbc"),
    (r"ab+c", "abbc"),
    (r"colou?r", "color"),
    (r"\d+", "Marks: 95"),
    (r"\w+", "Python_123"),
    (r"\s", "Hello World")
]

for pattern, text in examples:
    match = re.search(pattern, text)

    print(f"Pattern : {pattern}")
    print(f"Text    : {text}")
    print(f"Match   : {match.group() if match else 'No Match'}")
    print("-" * 45)
```
