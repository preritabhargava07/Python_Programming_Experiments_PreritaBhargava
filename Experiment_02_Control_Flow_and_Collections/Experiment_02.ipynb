## 1. Control Flow Statements

### 1(a) Conditional Statements

Conditional statements allow a program to make decisions based on specified conditions.

#### Simple `if` Statement
The `if` statement executes a block of code only when its condition is true.

```python
# Demonstration of Simple if Statement

temperature = 32

print("TEMPERATURE CHECK")
print("-" * 40)

print("Current Temperature:", temperature, "°C")

if temperature > 30:
    print("It's a hot day! Stay hydrated.")
```

#### `if-else` Statement

The `if-else` statement selects one of two blocks depending on whether the given condition is true or false.

```python
# Demonstration of if-else Statement

number = 27

print("EVEN OR ODD CHECK")
print("-" * 40)

print("Number:", number)

if number % 2 == 0:
    print(number, "is an Even Number.")
else:
    print(number, "is an Odd Number.")
```

#### Nested `if` Statement

A nested `if` statement places one conditional statement inside another to evaluate multiple levels of conditions.

```python
# Demonstration of Nested if Statement

age = 20
has_id = True

print("EVENT ENTRY CHECK")
print("-" * 40)

if age >= 18:
    print("Age requirement satisfied.")

    if has_id:
        print("Valid ID available.")
        print("Entry Approved!")
    else:
        print("Entry Denied: Valid ID required.")
else:
    print("Entry Denied: Minimum age is 18.")
```

### 1(b) Looping Statements
Loops are used to execute a block of code repeatedly.
#### ⁠ for ⁠ Loop
A ⁠ for ⁠ loop is used to iterate over a sequence or a specified range of values.


```python
# Demonstration of for Loop
print("MULTIPLICATION TABLE OF 7")
print("-" * 40)
for i in range(1, 11):
    print("7 x", i, "=", 7 * i)
```

#### ⁠ while ⁠ Loop
A ⁠ while ⁠ loop repeatedly executes a block of code as long as the given condition remains true.


```python
# Demonstration of while Loop
print("COUNTDOWN")
print("-" * 40)
count = 5
while count > 0:
    print(count)
    count -= 1
print("Let's Go!")
```

## 2. Python Collections
Python provides different collection data types for storing and managing multiple values.
### 2(a) Lists
A list is an *ordered and mutable* collection.  
The following program demonstrates ⁠ append() ⁠, ⁠ insert() ⁠, ⁠ remove() ⁠ and slicing.

```python
# List Creation and Manipulation
subjects = ["Python", "Physics", "Mathematics", "Economics"]
print("LIST OPERATIONS")
print("-" * 40)
print("Original List :", subjects)

# append()
subjects.append("English")
print("After Append  :", subjects)

# insert()
subjects.insert(1, "Calculus")
print("After Insert  :", subjects)

# remove()
subjects.remove("Economics")
print("After Remove  :", subjects)

# slicing
print("First 3 Items :", subjects[:3])
print("Last 2 Items  :", subjects[-2:])
```

### 2(b) Tuples
A tuple is an *ordered and immutable* collection.  
Its elements can be accessed using indexing, but they cannot be modified after creation.

```python
# Tuple Indexing and Immutability Demonstration
student = ("Prerita", 19, "Python", 92.5)
print("TUPLE DEMONSTRATION")
print("-" * 40)
print("Complete Tuple :", student)
print("Student Name   :", student[0])
print("Age            :", student[1])
print("Course         :", student[2])
print("Score          :", student[3])
print("\nTuples are immutable.")
print("The statement student[1] = 19 would produce a TypeError.")
```

### 2(c) Sets
A set is an *unordered collection of unique elements*.  
Sets support mathematical operations such as *union, intersection and difference*.

```python
# Demonstration of Set Operations
python_students = {"Aarav", "Riya", "Kabir", "Meera"}
ai_students = {"Riya", "Meera", "Arjun", "Sara"}
print("SET OPERATIONS")
print("-" * 40)
print("Python Students :", python_students)
print("AI Students     :", ai_students)
print("\nUnion        :", python_students | ai_students)
print("Intersection :", python_students & ai_students)
print("Difference   :", python_students - ai_students)
```

### 2(d) Dictionaries

A dictionary stores data in **key-value pairs**.  
The following program demonstrates adding, updating and deleting dictionary elements.

```python
# Dictionary Creation and Manipulation

student = {
    "name": "Neo",
    "roll_no": 101,
    "course": "Python",
    "marks": 88
}

print("DICTIONARY OPERATIONS")
print("-" * 40)

print("Original Record :", student)

# Adding a new element
student["attendance"] = 95
print("\nAfter Adding Attendance:")
print(student)

# Updating an existing element
student["marks"] = 93
print("\nAfter Updating Marks:")
print(student)

# Deleting an element
del student["course"]
print("\nAfter Deleting Course:")
print(student)
```

## 3. Real-Life Application of Collections

### 🎓 Student Record Management System

The following mini application combines **lists and dictionaries** to maintain student records, calculate averages and identify performance levels.

```python
# Mini Student Record Management System

students = [
    {"name": "Aarav", "roll_no": 101, "marks": [85, 90, 88]},
    {"name": "Riya", "roll_no": 102, "marks": [92, 95, 89]},
    {"name": "Kabir", "roll_no": 103, "marks": [72, 78, 75]}
]

print("STUDENT RECORD MANAGEMENT SYSTEM")
print("=" * 45)

for student in students:
    average = sum(student["marks"]) / len(student["marks"])

    if average >= 90:
        performance = "Excellent"
    elif average >= 75:
        performance = "Good"
    else:
        performance = "Needs Improvement"

    print("\nRoll No     :", student["roll_no"])
    print("Name        :", student["name"])
    print("Marks       :", student["marks"])
    print("Average     :", round(average, 2))
    print("Performance :", performance)
```
