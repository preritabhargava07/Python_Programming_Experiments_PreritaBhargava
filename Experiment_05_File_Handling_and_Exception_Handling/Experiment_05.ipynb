## 1. File Access Modes

Python provides different modes for accessing files.

| Mode | Description |
|---|---|
| `r` | Opens an existing file for reading. |
| `w` | Opens a file for writing, creating it if necessary and overwriting existing content. |
| `a` | Opens a file for appending without deleting existing content. |

### Program: Demonstration of Read, Write and Append Modes

The following program creates a student record file, adds a new record and reads the updated contents.

```python
# Experiment 5 - Question 1
# Demonstration of File Access Modes

from pathlib import Path

data_folder = Path("lab_files")
data_folder.mkdir(exist_ok=True)

file_path = data_folder / "student_records.txt"

print("FILE ACCESS MODES")
print("-" * 45)

# Write Mode (w)
with open(file_path, "w", encoding="utf-8") as file:
    file.write("Student: Neo | Marks: 92\n")

print("Write Mode  : Initial record saved.")

# Append Mode (a)
with open(file_path, "a", encoding="utf-8") as file:
    file.write("Student: Riya | Marks: 88\n")

print("Append Mode : New record added.")

# Read Mode (r)
with open(file_path, "r", encoding="utf-8") as file:
    content = file.read()

print("Read Mode   : File contents retrieved.")

print("\nSTUDENT RECORDS")
print("-" * 45)
print(content)
```

## 3. Directory Operations Using the os Module

The `os` module provides functions for interacting with the operating system, including creating, listing, renaming and deleting directories.

### Program: Directory Management System

**Aim:** To demonstrate common directory operations using Python's `os` module.

**Operations performed:**
1. Display the current working directory.
2. Create a new directory.
3. List directory contents.
4. Rename the directory.
5. Remove the directory.

```python
# Experiment 5 - Question 3
# Directory Operations Using os Module

import os

print("DIRECTORY MANAGEMENT SYSTEM")
print("-" * 45)

print("Current Directory:", os.getcwd())

# Create a directory
os.makedirs("Demo_Folder", exist_ok=True)
print("\nDirectory Created: Demo_Folder")

# List directory contents
print("\nDirectory Contents:")
print(sorted(os.listdir("Demo_Folder")))

# Rename the directory
if os.path.exists("Renamed_Demo_Folder"):
    os.rmdir("Renamed_Demo_Folder")

os.rename("Demo_Folder", "Renamed_Demo_Folder")
print("\nDirectory Renamed: Renamed_Demo_Folder")

# Remove the directory
os.rmdir("Renamed_Demo_Folder")
print("Directory Removed Successfully!")
```

## 4. Exception Handling

Exception handling allows a program to manage runtime errors without terminating unexpectedly.

### 4(a) try-except-else-finally

- **try:** Contains code that may raise an exception.
- **except:** Handles the exception.
- **else:** Executes when no exception occurs.
- **finally:** Executes whether an exception occurs or not.

**Program:** Safe Division Calculator.

```python
# Experiment 5 - Question 4(a)
# Demonstration of try-except-else-finally

print("SAFE DIVISION CALCULATOR")
print("-" * 45)

try:
    numerator = 100
    denominator = 5

    result = numerator / denominator

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

else:
    print("Division Successful!")
    print("Result:", result)

finally:
    print("Execution Completed.")
```

### 4(b) Handling Multiple Exceptions

Python allows multiple exception types to be handled using separate `except` blocks.

**Program:** Demonstrate handling `ValueError` and `ZeroDivisionError`.

```python
# Experiment 5 - Question 4(b)
# Handling Multiple Exceptions

def safe_division(a, b):
    try:
        numerator = float(a)
        denominator = float(b)
        result = numerator / denominator

    except ValueError:
        print("Error: Please enter valid numeric values.")

    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")

    else:
        print("Result:", result)


print("MULTIPLE EXCEPTION HANDLING")
print("-" * 45)

safe_division(20, 4)
safe_division(10, 0)
safe_division("abc", 5)
```

### 4(c) Demonstration of raise and assert

- **raise:** Used to explicitly generate an exception.
- **assert:** Used to verify that a condition is true and raises `AssertionError` if it is false.

**Program:** Validate student marks and verify a minimum passing score.

```python
# Experiment 5 - Question 4(c)
# Demonstration of raise and assert

def validate_marks(marks):
    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    return "Marks are valid."


print("RAISE AND ASSERT DEMONSTRATION")
print("-" * 45)

try:
    print(validate_marks(85))
    print(validate_marks(120))

except ValueError as error:
    print("Raised Exception:", error)


try:
    score = 35
    assert score >= 40, "Student has not achieved passing marks."
    print("Student Passed!")

except AssertionError as error:
    print("Assertion Failed:", error)
```

## 5. Custom Exception

A custom exception is a user-defined exception class created by inheriting from Python's built-in `Exception` class.

### Program: Student Attendance Eligibility System

**Aim:** To create and handle a custom exception when a student's attendance falls below the required percentage.

```python
# Experiment 5 - Question 5
# Custom Exception Demonstration

class LowAttendanceError(Exception):
    """Raised when attendance is below the required percentage."""
    pass


def check_attendance(attendance):
    if attendance < 75:
        raise LowAttendanceError(
            "Attendance is below the required 75%."
        )

    return "Student is eligible for examination."


print("STUDENT ATTENDANCE ELIGIBILITY")
print("-" * 45)

students = {
    "Aarav": 92,
    "Riya": 68,
    "Kabir": 81
}

for name, attendance in students.items():
    print(f"\nStudent: {name}")
    print(f"Attendance: {attendance}%")

    try:
        print(check_attendance(attendance))

    except LowAttendanceError as error:
        print("Eligibility Error:", error)
```
