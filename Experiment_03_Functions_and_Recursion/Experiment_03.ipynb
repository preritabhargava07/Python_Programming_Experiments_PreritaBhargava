## 1. User-Defined Functions

A function is a reusable block of code designed to perform a specific task.

Python uses the `def` keyword to define a function. Functions improve code readability, reusability and organization.

### Program 1: Student Grade Calculator

```python
# Demonstration of User-Defined Functions

def calculate_grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 75:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "F"


print("STUDENT GRADE CALCULATOR")
print("-" * 40)

student_name = "Neo"
student_marks = 87

grade = calculate_grade(student_marks)

print("Student Name :", student_name)
print("Marks        :", student_marks)
print("Grade        :", grade)
```

## 2. Types of Function Arguments

### 2(a) Default Arguments

Default arguments allow a function to use a predefined value when an argument is not explicitly provided.

### Program 2: Personalized Welcome Message

```python
# Demonstration of Default Arguments

def welcome_student(name, course="Python Programming"):
    print(f"Welcome, {name}!")
    print(f"Enrolled Course: {course}")
    print("-" * 40)


print("DEFAULT ARGUMENTS")
print("-" * 40)

welcome_student("Aarav")
welcome_student("Riya", "Artificial Intelligence")
```

### 2(b) Keyword Arguments

Keyword arguments allow values to be passed to a function using parameter names.

This makes function calls more readable and allows arguments to be supplied in any order.

### Program 3: Student Profile

```python
# Demonstration of Keyword Arguments

def display_profile(name, age, course):
    print("STUDENT PROFILE")
    print("-" * 40)
    print("Name   :", name)
    print("Age    :", age)
    print("Course :", course)


display_profile(
    course="Python Programming",
    name="Meera",
    age=19
)
```

### 2(c) Variable-Length Arguments

Variable-length arguments allow functions to accept a flexible number of inputs.

- `*args` collects positional arguments into a tuple.
- `**kwargs` collects keyword arguments into a dictionary.

### Program 4: Flexible Student Report

```python
# Demonstration of Variable-Length Arguments

def calculate_average(*marks):
    if not marks:
        return 0
    return sum(marks) / len(marks)


def student_details(**details):
    for key, value in details.items():
        print(f"{key.title()} : {value}")


print("FLEXIBLE STUDENT REPORT")
print("-" * 40)

print("Average Marks:", round(calculate_average(85, 90, 95), 2))

print("\nStudent Details:")
student_details(name="Kabir", age=18, course="Python")
```

## 3. Lambda Functions

### 3(a) Lambda Function with map()

A lambda function is a small anonymous function defined using the `lambda` keyword.

The `map()` function applies a function to every element of an iterable.

**Program:** Calculate the squares of numbers using `lambda` and `map()`.

```python
# Experiment 3 - Question 3(a)
# Lambda Function with map()

numbers = [2, 4, 6, 8, 10]

squares = list(map(lambda x: x ** 2, numbers))

print("LAMBDA WITH MAP()")
print("-" * 40)

print("Original Numbers :", numbers)
print("Squared Numbers  :", squares)
```

### 3(b) Lambda Function with filter()

The `filter()` function selects elements from an iterable that satisfy a given condition.

A `lambda` function can be used to define this condition in a single line.

**Program:** Filter students who have scored 75 marks or above.

```python
# Experiment 3 - Question 3(b)
# Lambda Function with filter()

marks = [45, 82, 67, 91, 73, 88, 56]

high_scores = list(filter(lambda x: x >= 75, marks))

print("LAMBDA WITH FILTER()")
print("-" * 40)

print("Original Marks :", marks)
print("Marks >= 75    :", high_scores)
```

## 4. Recursive Programs

Recursion is a programming technique in which a function calls itself to solve a problem.

A recursive function contains:
- **Base Case:** A condition that stops recursion.
- **Recursive Case:** A function call that reduces the problem into smaller subproblems.

### 4(a) Factorial Using Recursion

The factorial of a non-negative integer is the product of all positive integers up to that number.

**Formula:** n! = n × (n-1)!

**Program:** Calculate the factorial of a number using recursion.

```python
# Experiment 3 - Question 4(a)
# Factorial Using Recursion

def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


print("RECURSIVE FACTORIAL CALCULATOR")
print("-" * 40)

number = 5

if number < 0:
    print("Factorial is not defined for negative numbers.")
else:
    print("Number    :", number)
    print("Factorial :", factorial(number))
```

### 4(b) Fibonacci Series Using Recursion

The Fibonacci series is a sequence in which each number is the sum of the two preceding numbers.

The sequence begins with 0 and 1.

**Formula:** F(n) = F(n-1) + F(n-2)

**Program:** Generate the first 10 Fibonacci numbers using recursion.

```python
# Experiment 3 - Question 4(b)
# Fibonacci Series Using Recursion

def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)


print("RECURSIVE FIBONACCI SERIES")
print("-" * 40)

terms = 10

print("Number of Terms :", terms)
print("Fibonacci Series:")

for i in range(terms):
    print(fibonacci(i), end=" ")
print()
```

## 5. Application Using Modular Functions

### 🎓 Student Academic Performance Analyzer

**Aim:** To develop a small Python application using multiple user-defined functions to evaluate student performance.

**Features:**
- Calculate total marks.
- Calculate percentage.
- Determine the grade.
- Check pass/fail status.
- Display a formatted student report.

**Concept:** Modular programming divides a large program into smaller, reusable functions, making the code easier to understand, test and maintain.

```python
# Experiment 3 - Question 5
# Student Academic Performance Analyzer

def calculate_total(marks):
    return sum(marks)


def calculate_percentage(marks):
    return calculate_total(marks) / len(marks)


def determine_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 75:
        return "A"
    elif percentage >= 60:
        return "B"
    elif percentage >= 40:
        return "C"
    else:
        return "F"


def check_result(marks):
    if all(mark >= 40 for mark in marks):
        return "PASS"
    return "FAIL"


def display_report(name, marks):
    total = calculate_total(marks)
    percentage = calculate_percentage(marks)
    result = check_result(marks)
    grade = determine_grade(percentage) if result == "PASS" else "F"

    print("=" * 45)
    print("       STUDENT PERFORMANCE REPORT")
    print("=" * 45)

    print("Student Name :", name)
    print("Marks        :", marks)
    print("Total Marks  :", total, "/", len(marks) * 100)
    print("Percentage   :", round(percentage, 2), "%")
    print("Grade        :", grade)
    print("Result       :", result)

    print("=" * 45)


# Main Program
student_name = "Neo"
student_marks = [92, 85, 89, 94, 88]

display_report(student_name, student_marks)
```
